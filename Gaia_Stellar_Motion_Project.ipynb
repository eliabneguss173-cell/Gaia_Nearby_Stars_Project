```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a small Gaia-style nearby stars dataset directly in Colab
np.random.seed(42)

data = {
    "parallax": np.random.uniform(10, 50, 300),   # mas
    "pmra": np.random.normal(0, 80, 300),         # mas/year
    "pmdec": np.random.normal(0, 80, 300)         # mas/year
}

df = pd.DataFrame(data)

# Show the first few rows
df.head()

# Calculate distance and total proper motion
df["distance_pc"] = 1000 / df["parallax"]
df["proper_motion"] = np.sqrt(df["pmra"]**2 + df["pmdec"]**2)

# Plot
plt.scatter(df["distance_pc"], df["proper_motion"], s=10)
plt.xlabel("Distance (parsecs)")
plt.ylabel("Proper Motion (mas/year)")
plt.title("Proper Motion vs Distance (Nearby Stars)")
plt.show()
```
